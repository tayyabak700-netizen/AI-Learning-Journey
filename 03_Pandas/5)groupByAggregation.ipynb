```python
import numpy as np
import pandas as pd 
```

```python
data = {
    "Category": ["A", "B", "A", "B", "A", "B", "A", "B"],
    "Store": ["S1", "S1", "S2", "S2", "S1", "S2", "S2", "S1"],
    "Sales": [100, 200, 150, 250, 120, 180, 200, 300],
    "Quantity":[10, 15, 12, 18, 8, 20, 15, 25],
    "Date": pd.date_range("2023-01-01", periods=8),
}
df = pd.DataFrame(data)
```

```python
df
```

```python
#Group by catagories and calculate the sum oof sales
cat = df.groupby("Category")['Sales'].sum()
cat
```

```python
#Group by stores and calculate the sum of sales
stores = df.groupby("Store")['Sales'].sum()
stores
```

```python
#Group by multiple columns
#Group by category and Storr
catStor = df.groupby(['Category', 'Store'])['Sales'].sum()
catStor
```
