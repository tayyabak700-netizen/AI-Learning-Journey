```python
# Which country has the highest popuation 
df[df['population']  == df['population'].max()]
```

```python
# Which is the capital of the country with highest population
df[df['population'] == df['population'].max()]['capital_city']
```

```python
# Which county has the least population 
df[df['population'] == df['population'].min()]['country']
```

```python
# What is the capital of the country with least population 
df[df['population'] == df['population'].min()]['capital_city']
```

```python
# Give me the top 5 countries with highest democratic score 
df.sort_values(by='democracy_score',ascending=False,inplace=True)
```

```python
# How many toltal reigions are there 
df['region'].value_counts().count()
```

```python
# How  many countries lie in Estern Europe region 
#df['region'].value_counts()['Eastern Europe']
df[df['region'] == "Eastern Europe"]['country']
```

```python
# Who is political leader of the 2nd highest population country  
df[df['population'] == df['population'].nlargest(2).iloc[1]]['political_leader']
```

```python
# How many countries are there whose political leaders are unknown 
df[df['political_leader'].isna()]['country'].count()
```

```python
# How many countries have Republic in their full name

count = 0 
def counting(txt):
    global count
    if 'republic' in txt.lower(): 
        count += 1
    return txt 
df['country_long'].apply(counting)
print(count)
```
