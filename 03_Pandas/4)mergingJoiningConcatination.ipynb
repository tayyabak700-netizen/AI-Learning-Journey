```python
import numpy as np 
import pandas as pd
```

**MERGING DATAFRAMES**

```python
# DataFrame 1: Employee Details
employees = pd.DataFrame({
    'employee_id': [101, 102, 103, 104],
    'name': ['Alice', 'Bob', 'Charlie', 'David'],
    'department': ['IT', 'HR', 'IT', 'Finance']
})
# DataFrame 2: Employee Salaries
salaries = pd.DataFrame({
    'employee_id': [101, 102, 103, 105],
    'salary': [85000, 70000, 90000, 95000]
})
```

```python
employees
```

```python
salaries
```

```python
pd.merge(employees, salaries)
```

```python
pd.merge(employees, salaries, on="employee_id", how = "inner")
```

```python
pd.merge(employees, salaries, on="employee_id", how = "outer")
```

```python
pd.merge(employees, salaries, on="employee_id", how = "left")
```

```python
pd.merge(employees, salaries, on="employee_id", how = "right")
```

**CONCATINATION OF 2 DATAFRAMES**

```python
# DataFrames from the image
df1 = pd.DataFrame(
    {
        "A": ["A0", "A1", "A2"],
        "B": ["B0", "B1", "B2"],
        "C": ["C0", "C1", "C2"],
    }
)

df2 = pd.DataFrame(
    {
        "A": ["A3", "A4", "A5"],
        "B": ["B3", "B4", "B5"],
        "C": ["C3", "C4", "C5"],
    }
)
```

```python
df1
```

```python
df2
```

```python
#Cocatination on the basis of olumns
pd.concat([df1,df2])
```

```python
#Concatination on the basis of rpow
pd.concat([df1,df2], axis=1)
```

**JOINING OF 2 DATAFRAMES**

```python
df1 = pd.DataFrame({
    "name": ["Alice", "Bob", "Charlie"]
}, index=[1, 2, 3])

df2 = pd.DataFrame({
    "score": [85, 90, 75]
}, index=[2, 3, 4])
```

```python
df1
```

```python
df2
```

```python
df1.join(df2)
```

```python
df2.join(df1)
```

```python
df1.join(df2, how='outer')
```
