```python
def extract_episodes(txt):
    check = False
    data = ""
    for i in txt:
        if i == ")":
            check = False
            return data
        if check == True:
            data = data + i
        if i == "(":
            check = True
        
```

```python
df['Episodes'] = df["Title"].apply(extract_episodes)
df.loc[4]['Episodes']
```

```python
# Remove Extra space and eps
df['Episodes']  = df['Episodes'].str.replace(" eps","")
```

```python
#Convert Episode in integer 
df['Episodes'] = df['Episodes'].astype(int)
```

```python
def extract_time(txt):
    check = False
    data = ""
    for i in range(len(txt)):
        if txt[i] == ')':
            for j in range(i+1, i+20):
                data += txt[j]
                
            return data
```

```python
df['Total Time'] = df['Title'].apply(extract_time)
```

```python
df.head()
```

```python
# 1. Split the 'Total Time' string into temporary 'start' and 'end' strings
split_dates = df['Total Time'].str.split(' - ', expand=True)

# 2. Convert both columns to pandas datetime objects
start_dt = pd.to_datetime(split_dates[0], format='%b %Y')
end_dt = pd.to_datetime(split_dates[1], format='%b %Y')

# 3. Calculate the difference in months
df['Total Months'] = ((end_dt.dt.year - start_dt.dt.year) * 12 + 
                      (end_dt.dt.month - start_dt.dt.month) + 1)
```

```python
# Find the row with the maximum number of months
longest_running_anime = df.loc[df['Total Months'].idxmax()]

# Display the title and the duration
print(f"Longest running anime: {longest_running_anime ['Title']}")
print(f"Duration: {longest_running_anime['Total Months']} months")
```
